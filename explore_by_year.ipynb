```python
import tempfile
import pandas as pd
import urllib.request
import os

# ----------------------------- weather by year X dataset --------------------------------

# raw csv had no header row.
cols = ['STATION', 'DATE', 'ELEMENT', 'VALUE', 'M_FLAG', 'Q_FLAG', 'S_FLAG', 'OBS_TIME']
url = "https://www.ncei.noaa.gov/pub/data/ghcn/daily/by_year/2015.csv.gz"   # CHANGE 1763 to whatever year to explore

# Note this (1 years worth of data) takes like a minute to load
with tempfile.NamedTemporaryFile(suffix=".csv.gz", delete=False) as tmp:   
    urllib.request.urlretrieve(url, tmp.name)
    weather_yearX = pd.read_csv(tmp.name, header=None, names=cols, compression='gzip')

print(f'temp file location: {tmp.name}')
os.remove(tmp.name)    # cleanup handled here


```

```python
# ----------------------------- station dataset --------------------------------

url = "https://www.ncei.noaa.gov/pub/data/ghcn/daily/ghcnd-stations.txt"
colwidths = [(0,11), (12,20), (21,30), (31,37), (38,40), (41,71), (72,75), (76,79), (80,85)]
cols = ['ID', 'LATITUDE', 'LONGITUDE', 'ELEVATION', 'STATE', 'NAME', 'GSN FLAG', 'HCN/CRN FLAG', 'WMO ID']

with tempfile.NamedTemporaryFile(suffix=".txt", delete=False) as tmp:
    urllib.request.urlretrieve(url, tmp.name)
    stations = pd.read_fwf(tmp.name, colspecs=colwidths, header=None, names=cols)

print(f'temp file location: {tmp.name}')
os.remove(tmp.name)
```

```python
merged = weather_yearX.merge(stations, left_on='STATION', right_on='ID', how='left')
```
